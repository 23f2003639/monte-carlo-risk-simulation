```python
import pandas as pd

# File names for all five stocks
stock_files = {
    "RELIANCE": "RELIANCE.csv",
    "TCS": "TCS.csv",
    "ICICIBANK": "ICICIBANK.csv",
    "BHARTIARTL": "BHARTIARTL.csv",
    "LT": "L&T.csv"
}

# Read all CSV files
stock_data = {}

for stock, file in stock_files.items():
    stock_data[stock] = pd.read_csv(file, encoding="utf-8-sig")
    print(f"{stock} data loaded successfully!")
    display(stock_data[stock].head())
```

**Clean Column Names and Keep Only EQ Series**

```python
for stock, df in stock_data.items():

    # Remove leading and trailing spaces from column names
    df.columns = df.columns.str.strip()

    # Remove spaces from Series values
    df["Series"] = df["Series"].str.strip()

    # Keep only equity (EQ) rows
    df = df[df["Series"] == "EQ"].copy()

    # Update the cleaned data
    stock_data[stock] = df

    print(f"{stock}: {len(df)} EQ rows retained")
```

**Convert Dates and Closing Prices**

```python
for stock, df in stock_data.items():

    # Convert Date column to datetime format
    df["Date"] = pd.to_datetime(
        df["Date"],
        format="%d-%b-%Y"
    )

    # Remove commas and convert Close Price to float
    df["Close Price"] = (
        df["Close Price"]
        .str.replace(",", "", regex=False)
        .astype(float)
    )

    # Update the cleaned data
    stock_data[stock] = df

    print(f"{stock}: Date and Close Price converted successfully")

# Display a sample to verify the conversions
display(stock_data["RELIANCE"][["Date", "Close Price"]].head())
```

**Sort Dates and Combine Closing Prices**

```python
import numpy as np
import pandas as pd

series_list = {}
for stock, df in stock_data.items():
    df = df.copy()
    if "Series" in df.columns:
        df = df[df["Series"].str.strip() == "EQ"]       # keep only normal equity rows
    df = df.sort_values("Date").set_index("Date")
    series_list[stock] = df["Close Price"]

prices = pd.concat(series_list, axis=1)
prices = prices.rename(columns={"L_T": "LT", "L&T": "LT"})   # fix L&T name if needed
raw_prices = prices.copy()                                   # untouched copy

print(prices.shape)
prices.head()
```

```python
events = pd.DataFrame([
    ("RELIANCE",   "2023-07-20", "demerger"),
    ("RELIANCE",   "2024-10-28", "bonus"),
    ("BHARTIARTL", "2021-09-27", "rights"),
], columns=["ticker", "ex_date", "type"])

def check_events(events, price_table):
    rets = price_table.pct_change()
    vol = rets.std()
    rows = []
    for _, e in events.iterrows():
        d = pd.Timestamp(e["ex_date"])
        if d not in rets.index:                              # if not a trading day,
            d = rets.index[rets.index.searchsorted(d)]       # use the next trading day
        r = rets.loc[d, e["ticker"]]
        rows.append({"ticker": e["ticker"], "ex_date": d.date(), "type": e["type"],
                     "return": round(r, 4),
                     "z_score": round(r / vol[e["ticker"]], 1)})
    return pd.DataFrame(rows)

print("BEFORE adjusting")
check_events(events, raw_prices)
```

```python
# Factor rules:
#   Bonus (a new for every b held) -> b / (a + b)       1:1 bonus -> 0.5
#   Split (old FV -> new FV)       -> new / old          10 -> 2 gives 0.2
#   Rights issue                   -> TERP / cum-rights close
#   Demerger                       -> ex-price / previous close

# Airtel rights factor (1 new share for every 14 held, issue price 535)
P = raw_prices.loc[raw_prices.index < "2021-09-27", "BHARTIARTL"].iloc[-1]   # last close before ex-date
f_rights = ((14 * P + 535) / 15) / P
print("Airtel cum-rights close:", P, "| factor:", round(f_rights, 5))        # about 739.4 and 0.9816

adjustments = [
    ("RELIANCE",   "2023-07-20", 2580 / 2841.85),   # demerger (verify 2580 in the NSE circular)
    ("RELIANCE",   "2024-10-28", 0.5),              # 1:1 bonus
    ("BHARTIARTL", "2021-09-27", f_rights),         # rights issue
]

adj_prices = raw_prices.copy()          # always start fresh, so re-running is safe
for ticker, ex_date, f in adjustments:
    adj_prices.loc[adj_prices.index < ex_date, ticker] *= f
```

## **PHASE 5**

### **The Gaussian Monte Carlo model**

**Settings and parameters (from TRAIN only)**

```python
np.random.seed(42)                      # makes results repeatable

n_sims   = 10_000                       # number of simulated futures
n_days   = 252                          # one trading year
n_assets = 5
init     = 1_000_000                    # Rs 10 lakh
weights  = np.full(n_assets, 0.20)      # equal weight, buy-and-hold

mu  = train.mean().values               # average daily log return, shape (5,)
cov = train.cov().values                # daily covariance matrix, shape (5, 5)

print("Annualised mean:", (mu * 252).round(3))
print("Annualised vol :", (np.sqrt(np.diag(cov)) * np.sqrt(252)).round(3))
```

**Simulate**

```python
L = np.linalg.cholesky(cov)                                  # cov = L @ L.T

Z = np.random.standard_normal((n_sims, n_days, n_assets))   # independent random shocks
daily = mu + Z @ L.T                                         # correlated daily log returns

# Value of each stock's holding over time (log returns add up, then exp)
stock_values = (init * weights) * np.exp(daily.cumsum(axis=1))
port_paths = stock_values.sum(axis=2)                                 # (10000, 252)
port_paths = np.hstack([np.full((n_sims, 1), init), port_paths])      # add day 0

print(port_paths.shape)                                      # (10000, 253)
```

**Sanity check (does the simulation match the data?)**

```python
sim_cov  = np.cov(daily.reshape(-1, n_assets).T)
print("Max covariance difference:", np.abs(sim_cov - cov).max())     # tiny, about 5e-7

sim_corr = pd.DataFrame(np.corrcoef(daily.reshape(-1, n_assets).T),
                        index=train.columns, columns=train.columns)
print("Simulated correlations:\n", sim_corr.round(2))
print("Train correlations:\n", train.corr().round(2))
```

**Risk metrics (reusable for the bootstrap model later)**

```python
def risk_summary(paths, init):
    final_ret = paths[:, -1] / init - 1          # 1-year return of each path
    day1_ret  = paths[:, 1] / init - 1           # first-day return of each path

    var95  = -np.percentile(final_ret, 5)
    var99  = -np.percentile(final_ret, 1)
    cvar95 = -final_ret[final_ret <= -var95].mean()
    cvar99 = -final_ret[final_ret <= -var99].mean()

    d_var95  = -np.percentile(day1_ret, 5)
    d_var99  = -np.percentile(day1_ret, 1)
    d_cvar95 = -day1_ret[day1_ret <= -d_var95].mean()
    d_cvar99 = -day1_ret[day1_ret <= -d_var99].mean()

    running_max = np.maximum.accumulate(paths, axis=1)
    max_dd = ((paths - running_max) / running_max).min(axis=1)

    return pd.Series({
        "Mean 1y return":        final_ret.mean(),
        "Median 1y return":      np.median(final_ret),
        "Std of 1y return":      final_ret.std(),
        "Probability of loss":   (final_ret < 0).mean(),
        "1y VaR 95%":            var95,
        "1y CVaR 95%":           cvar95,
        "1y VaR 99%":            var99,
        "1y CVaR 99%":           cvar99,
        "1d VaR 95%":            d_var95,
        "1d CVaR 95%":           d_cvar95,
        "1d VaR 99%":            d_var99,
        "1d CVaR 99%":           d_cvar99,
        "Median max drawdown":   np.median(max_dd),
        "Worst-5% max drawdown": np.percentile(max_dd, 5),
    })

results = {}
results["Gaussian"] = risk_summary(port_paths, init)
pd.DataFrame(results).round(4)
```

## **PHASE 6**

### **The Bootstrap Model**

**Helper and simple bootstrap**

```python
X = train.values                 # training returns as a plain array, shape (990, 5)
T = len(X)

def to_paths(daily):
    """Turn simulated daily log returns (n_sims, n_days, 5) into portfolio value paths."""
    stock_values = (init * weights) * np.exp(daily.cumsum(axis=1))
    p = stock_values.sum(axis=2)
    return np.hstack([np.full((daily.shape[0], 1), init), p])

np.random.seed(42)
idx = np.random.randint(0, T, size=(n_sims, n_days))   # random day numbers
boot_daily = X[idx]                                     # (10000, 252, 5): whole rows, so correlations are kept

paths = {"Gaussian": port_paths,                        # from Cell 16
         "Bootstrap": to_paths(boot_daily)}
```

**Sanity check**

```python
boot_corr = np.corrcoef(boot_daily.reshape(-1, 5).T)
print("Max correlation difference vs train:", np.abs(boot_corr - train.corr().values).max().round(4))   # about 0.002
print("Worst day in train: ", X.mean(axis=1).min().round(4))
print("Worst day possible in bootstrap:", boot_daily.mean(axis=2).min().round(4))   # can't be worse than real history
```

**Block bootstrap**

```python
def block_bootstrap(data, block_len, n_sims, n_days, seed=42):
    np.random.seed(seed)
    T = len(data)
    n_blocks = int(np.ceil(n_days / block_len))
    starts = np.random.randint(0, T - block_len + 1, size=(n_sims, n_blocks))        # random block start days
    idx = (starts[:, :, None] + np.arange(block_len)[None, None, :]).reshape(n_sims, -1)[:, :n_days]
    return data[idx]                                                                  # consecutive days inside each block

for bl in (5, 10, 20):
    paths[f"Block {bl}d"] = to_paths(block_bootstrap(X, bl, n_sims, n_days))
```

**Comparison table**

```python
results = {name: risk_summary(p, init) for name, p in paths.items()}
comparison = pd.DataFrame(results)
comparison.round(4)
```

**Charts**

```python
fig, ax = plt.subplots(1, 2, figsize=(15, 5))

for name in ["Gaussian", "Bootstrap", "Block 10d"]:
    ax[0].hist(paths[name][:, -1] / init - 1, bins=80, histtype="step", density=True, lw=1.8, label=name)
ax[0].set_title("1-year return distribution")
ax[0].set_xlabel("Return"); ax[0].legend()

comparison.loc[["1d VaR 99%", "1d CVaR 99%"], ["Gaussian", "Bootstrap", "Block 10d"]].T.plot(kind="bar", ax=ax[1])
ax[1].set_title("1-day 99% risk by model")
ax[1].set_ylabel("Loss (fraction of portfolio)")
ax[1].tick_params(axis="x", rotation=0)

plt.tight_layout()
plt.show()
```

```python
def plot_model(paths_arr, name, ylim=(7, 19), xlim=(6, 24)):
    days = np.arange(paths_arr.shape[1])
    final_values = paths_arr[:, -1]
    var95 = -np.percentile(final_values / init - 1, 5)

    fig, ax = plt.subplots(1, 2, figsize=(15, 5))

    # Left: fan chart
    ax[0].plot(days, paths_arr[:20].T / 1e5, lw=0.5, color="grey", alpha=0.5)
    ax[0].fill_between(days, np.percentile(paths_arr, 5, axis=0) / 1e5,
                             np.percentile(paths_arr, 95, axis=0) / 1e5,
                       alpha=0.2, color="tab:blue", label="5-95%")
    ax[0].fill_between(days, np.percentile(paths_arr, 25, axis=0) / 1e5,
                             np.percentile(paths_arr, 75, axis=0) / 1e5,
                       alpha=0.35, color="tab:blue", label="25-75%")
    ax[0].plot(days, np.median(paths_arr, axis=0) / 1e5, color="navy", label="Median")
    ax[0].set_ylim(*ylim)
    ax[0].set_title(f"{name}: simulated portfolio value")
    ax[0].set_xlabel("Trading day"); ax[0].set_ylabel("Value (Rs lakh)"); ax[0].legend()

    # Right: distribution of final values
    ax[1].hist(final_values / 1e5, bins=60, alpha=0.7)
    ax[1].axvline(init / 1e5, color="black", label="Start (Rs 10 lakh)")
    ax[1].axvline(init * (1 - var95) / 1e5, color="red", ls="--", label="95% VaR level")
    ax[1].set_xlim(*xlim)
    ax[1].set_title(f"{name}: value after 1 year")
    ax[1].set_xlabel("Value (Rs lakh)"); ax[1].legend()

    plt.tight_layout()
    plt.show()
```

## **PHASE 7: Backtest on 2025**

**The actual 2025 portfolio returns**

```python
# Equal-weight portfolio, rebalanced to 20% each day (same thing the 1-day VaR describes)
test_port  = (np.exp(test)  - 1).mean(axis=1)
train_port = (np.exp(train) - 1).mean(axis=1)

print("Test days:", len(test_port))
print("Daily volatility  train: %.4f | test: %.4f" % (train_port.std(), test_port.std()))
print("Worst test day:", test_port.idxmin().date(), round(test_port.min(), 4))
```

**The backtest functions**

```python
from scipy import stats

def simulate_daily(train_df, n_sims=10_000, n_days=252, block_len=10, seed=42):
    """Simulated daily log returns from all three models, using only train_df."""
    X = train_df.values
    T = len(X)
    mu, cov = X.mean(axis=0), np.cov(X.T)

    np.random.seed(seed)
    Z = np.random.standard_normal((n_sims, n_days, 5))
    gaussian = mu + Z @ np.linalg.cholesky(cov).T

    np.random.seed(seed)
    bootstrap = X[np.random.randint(0, T, size=(n_sims, n_days))]

    np.random.seed(seed)
    n_blocks = int(np.ceil(n_days / block_len))
    starts = np.random.randint(0, T - block_len + 1, size=(n_sims, n_blocks))
    idx = (starts[:, :, None] + np.arange(block_len)[None, None, :]).reshape(n_sims, -1)[:, :n_days]
    block = X[idx]

    return {"Gaussian": gaussian, "Bootstrap": bootstrap, "Block 10d": block}


def kupiec(breaches, N, p):
    """Kupiec proportion-of-failures test. Returns (LR statistic, p-value)."""
    def loglik(q):
        if q == 0: return 0.0 if breaches == 0 else -np.inf
        if q == 1: return 0.0 if breaches == N else -np.inf
        return (N - breaches) * np.log(1 - q) + breaches * np.log(q)
    lr = -2 * (loglik(p) - loglik(breaches / N))
    return lr, stats.chi2.sf(lr, 1)


def backtest(train_df, test_df):
    test_p = (np.exp(test_df) - 1).mean(axis=1)           # actual daily portfolio returns
    N = len(test_p)
    rows = []
    for name, daily_arr in simulate_daily(train_df).items():
        sim_p = (np.exp(daily_arr) - 1).mean(axis=2).ravel()    # every simulated day, about 2.5 million
        for conf in (95, 99):
            a = 1 - conf / 100
            var  = -np.percentile(sim_p, a * 100)
            cvar = -sim_p[sim_p <= -var].mean()
            breach = test_p < -var
            x = int(breach.sum())
            lr, pval = kupiec(x, N, a)
            rows.append({"model": name, "conf": conf, "VaR": var, "CVaR": cvar,
                         "expected_breaches": a * N, "breaches": x,
                         "LR": lr, "p_value": pval,
                         "avg_loss_on_breach": -test_p[breach].mean() if x > 0 else np.nan})
    return pd.DataFrame(rows), test_p
```

```python
bt, test_p = backtest(train, test)
bt.round(4)
```

**Which days were the breaches?**

```python
for name in ["Gaussian", "Bootstrap"]:
    for conf in (95, 99):
        v = bt.loc[(bt.model == name) & (bt.conf == conf), "VaR"].iloc[0]
        b = test_p[test_p < -v]
        print(f"\n{name} {conf}% (VaR {v:.4f}): {len(b)} breaches")
        print(b.round(4).to_string())
```
